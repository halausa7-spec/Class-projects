#New Concept: Conditional Logic using IF statement

The idea is to execute code only is specific situations.
The situation is controlled by asking the computer to perform a test where the result of the test is either **True** or **False**.
<center>
<img src="https://upload.wikimedia.org/wikipedia/commons/thumb/c/c5/If-Then-Else-diagram.svg/640px-If-Then-Else-diagram.svg.png" width="300" />
</center>

Code in Python looks like this:
```
if (A) :
    B
else :
    C
```




````python
# Task 6 IF statement (conditional logic)
F=98.6
F=F+1.4
if (F>100)or (F==100):
  print("sick")
else:
  print("healthy")
print("health check complete")
````

#Comparison expressions & Operators
Compare two values and determine True or False

|Comparison Operators| Function |
|:-------------:|:----------|
|>|     Greater than|
|>=|   Greater than or equal
|==|    Equal|
|<=|    Less than or equal|
|<|     Less than|
|!=|    Not equal|



#Boolean (Logical)  Operators
**Boolean expressions** use logical operators to calculate a Boolean value. 'Logical' and 'Boolean' are synonyms in this context.

|Logical Operators| Function |
|:------------:|----------|
|and|  True if all statements are True|
|or|   True if any statements are True|
|not|   Reverses the result|


````python
u7=4
v7=6
if (u7<v7):
  print(u7,"<", v7)

# Adding tests for other comparison operators
if (u7 <= v7):
  print(u7, "<=", v7)

if (u7 > v7):
  print(u7, ">", v7)

if (u7 >= v7):
  print(u7, ">=", v7)

if (u7 == v7):
  print(u7, "==", v7)

if (u7 != v7):
  print(u7, "!=", v7)
````

````python
# Task 9 Blood Type
# In task 9 cell build a decision tree using nested IF statements to test each blood type and identify universal donors and universal recipients. Do this without defining any functions.

blood_type= "AB+" # Example: "+", "-"

print(f"Analyzing blood type: {blood_type}")
if blood_type == "O-":
         print("Identification: Universal Donor")
         print('Can donate to: All blood types (A+, A-, B+, B-, AB+, AB-, O+, O-)')
         print('Can receive from: O-')
else:
  if blood_type == "AB+":
    print("Identification: Universal Recipient")
    print('Can donate to: AB+')
    print('Can receive from: All blood types (A+, A-, B+, B-, AB+, AB-, O+, O-)')
  else:
    print("Identification: Neither Universal Recipient nor Universal Donor")
    if blood_type.endswith("+"):
        print(f"Blood type {blood_type} is Rh positive.")
    elif blood_type.endswith("-"):
        print(f"Blood type {blood_type} is Rh negative.")
    else:
        print("Invalid blood type format or unknown type.")
print('\nDecision tree complete.')
````
