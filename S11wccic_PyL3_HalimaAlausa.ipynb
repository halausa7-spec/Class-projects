#New Concept: FOR loop
Loop to execute a code-block multiple times

```
for loop-variable in sequence:
    code-block
```


**Sequence** is a string, list or range. **Sequence** determines the loop exit condition; the loop stops when the **sequence** runs out of data.

**Loop-variable** is the current item of data from **sequence**.

A **code-block** is code that will be executed in a loop.
Python indents **code-blocks** 4 spaces.



* Range

    range(3)
    
    is the numbers 0, 1, 2
    These are 3 items (numbers) in this range
    Loop exits after executing code-block 3 times
* String

    "hello world"

    there are 11 items (characters) in this string
    Loop exits after executing code-block 11 times

* List

    ['baby', 'mommy', 'daddy', 'grandma', 'grandpa']

    there are 5 items (text messages) in this list
    Loop exists after executing code-block 5 times

* Blocks of code appear after python commands that end with \:

* All lines in the block must have the same indent
* Typically a 4-space indent

#Build a Caesar Cypher Algorithm
First setup important varibles:

pta = Plain text alphabet, starts with a space

pos = position to rotate the alphabet

cta = rotated Cypher alphabet

````python
#Task 8a: Setup
pta=" " +alph
r=13
cta=" "+alph[r:]+alph[0:r]
print(pta)
print(cta)
````

````python
#Task 8b: Caesar cypher

# What message to work with
msg = "hello world"
msg = "uryyb jbeyq"
msg = "ebovafba pbyyrtr bs ohfvarff vf cneg bs trbetvn fgngr havirefvgl"   #13
msg = "kvmvhjpio kgpn viy rzgg nzz" #5

ccm=""

# Loop for every character of the message
for letter in msg:
    # Find position of current letter
    pos = pta.find( letter )
    # corresponding cypher letter
    cy = cta[pos]
    ccm += cy
# print output message
print(ccm)
````

#New Concept: Define a function

```
def function_name( arguments ):
    code-block
    return result
```


````python
#Task 8c: Caesar Cypher function
def cypher ( m, k):
  pta = " "+ alph
  cta = " " + alph[k:] + alph[0:k]
  cy=" "
  for letter in m:
    pos= pta.find(letter)
    cy += cta[pos]
  return cy

cypher(msg, 5)
````

````python
#Task 8d: testing the cypher function

msg = "kvmvhjpio kgpn viy rzgg nzz"; rot = 5
msg = "iwt sdv pit bn wdbtldgz"; rot = 11
#msg = "lbq lxoxg bl vhhdxw"; rot = 7
cypher(msg, rot)
````
