# Variables
**Variables** are a place in memory to store data. Spreadsheets assign names to places of memory, like A1 and B7 (cells in the spreadsheet grid). In Python you decide the name for each place of memory.

An **assignment** (the = sign) moves data to a named memory location. Data moves right to left into the variable. The first time you assign a variable you are creating that place of memory.

Python keeps track of spelling and upper/lower case on variable names. Spell it different and its a different place of memory.



Use the **{}** ***Variables*** control at bottom of screen to reveal the variables inspector.
